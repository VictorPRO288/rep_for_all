```python
import numpy as np
import pandas as pd

df = pd.read_csv("dirty_cafe_sales.csv")

dft = df.copy()

dft=dft.replace(['UNKNOWN', 'ERROR'], np.nan)
dft[['Quantity', 'Price Per Unit', 'Total Spent']] = dft[['Quantity', 'Price Per Unit', 'Total Spent']].astype(float)
dft['Price Per Unit'] = dft['Price Per Unit'].fillna(dft['Total Spent']/dft['Quantity'])
dft['Quantity'] = dft['Quantity'].fillna(dft['Total Spent']/dft['Price Per Unit'])
dft['Total Spent'] = dft['Total Spent'].fillna(dft['Price Per Unit']*dft['Quantity'])

item_price = {
    'Tea':1.5,
    'Coffee':2.0,
    'Cake':3.0,
    'Cookie':1.0,
    'Salad':5.0,
    'Smoothie':4.0,
    'Sandwich':4.0,
    'Juice':3.0
}

price_item={price: item for item, price in item_price.items()}
dft['Item'] = dft['Item'].fillna(dft['Price Per Unit'].map(price_item))    
dft['Price Per Unit'] = dft['Price Per Unit'].fillna(dft['Item'].map(item_price)) 
dft['Quantity'] = dft['Quantity'].fillna(dft['Total Spent']/dft['Price Per Unit'])
dft['Total Spent'] = dft['Total Spent'].fillna(dft['Price Per Unit']*dft['Quantity'])
dft=dft.dropna(subset=['Item'])
dft=dft.dropna(subset=['Transaction Date'])
# print(dft.sample(5))
print(dft.info())

# print(dft[dft['Quantity'].isnull()])



dft.to_csv('cleaned_cafe.csv')
```
