```python
import pandas as pd
import numpy as np

df = pd.read_csv('dirty_wiki.csv')

dft = df.copy()

dft = dft.drop(['Peak', 'All Time Peak', 'Ref.'], axis=1)
dft['Adjusted_gross_2022'] = dft['Adjusted_gross_2022'].str.replace(r'[$,\[\]a-zA-Z]', '', regex =True)
dft['Actual_gross'] = dft['Actual_gross'].str.replace(r'[$,\[\]a-zA-Z]', '', regex=True)
dft['Average_gross'] = dft['Average_gross'].str.replace(r'[$,\[\]a-zA-Z]', '', regex=True)
dft['Tour title'] = dft['Tour title'].str.replace(r'\[[^\]]*\]', '', regex=True)
dft['Tour title'] = dft['Tour title'].str.replace(r'[^\s\w&-]', '', regex=True)
dft[['Adjusted_gross_2022', 'Actual_gross', 'Average_gross']] = dft[['Adjusted_gross_2022', 'Actual_gross', 'Average_gross']].astype('Int64')

print(dft.info())
dft
```
